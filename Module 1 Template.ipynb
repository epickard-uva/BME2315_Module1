## Disease Background: 

* Prevalence & incidence: The prevalence of a disease is the number of people currently living with the disease. The prevalence of Alzheimer's disease is over 55 million worldwide. The incidence rate is the number of new cases each year. The incidence of Alzheimer's disease is around 10 million worldwide. In the US, prevalence is around 7.4 million people, and incidence is around 1 million for individuals who are 65 years old and up. 

* Economic burden: Alzheimer’s is one of the most economically costly industries globally and it is expected to reach $2 trillion by 2030. The cost of care per capita for Alzheimer’s patients and their families can range anywhere from $468 to $171,283 per year. Cost tends to progress nonlinearly as patients continue to degrade and require more severe and around-the-clock care. This disease exerts a huge economic burden not only on patients but on caregivers as well.  

* Risk factors(genetics & lifestyle): There are a variety of risk factors that influence somebody’s likelihood of developing Alzheimer's disease, including genetics, behaviors, habits, age, and family history. The largest factor is age, as the majority of people affected are 65 years or older. Having a direct family member with the disease also makes it more likely for it to develop. In terms of genetics, there are two types of genes that contribute to the development of the disease: deterministic and risk.  The Deterministic-type genes cause the disease but are responsible for less than 1% of cases. Some other factors include head injuries and heart problems such as diabetes or high blood pressure. Preventative lifestyle measures can also reduce risk, such as cognitive exercises, regular exercise, a healthy diet, not smoking, maintaining a healthy weight, and getting good sleep.  

* Societal determinants: There has been shown to be a correlation between a lower level of education and a higher risk for Alzheimer’s due to a lack of cognitive reserve. Those without high school degrees have a risk of almost twice as high compared to those who have completed high school and/or continued higher education. Additionally, factors such as access to healthcare, physical activity, healthy eating, and even social isolation can lead to decreased brain health and the development of brain degeneration. 

* Symptoms: The main symptom of Alzheimer's disease is memory loss that significantly impacts your daily life. Additional symptoms include difficulty planning, doing calculations, and concentrating on tasks. People’s perception of time and place can be altered as well. Mood changes may occur along with withdrawal from typical activities. Some other symptoms involve judgment becoming poor, trouble with speaking or conversating, and decreased ability to keep track of things.  A couple more symptoms involve vision being altered and judgment becoming poor. 

* Diagnosis: Alzheimer’s disease is diagnosed through a series of definitive tests and screenings done by healthcare professionals. Doctors strive to get a comprehensive understanding of the patients health by discussion of family history, symptoms, consultations with family/friends, lab tests, brain imaging, or memory tests, among others. Many early symptoms of Alzheimer’s present as cognitive decline, or a regression in memory sharpness, trouble with concentration or problem solving, and poor judgment. 

* Standard of care treatments (& reimbursement): Treatments of Alzheimer's involve different factors, including pharmacological treatment and non-pharmacological treatments. In terms of pharmacological options, there are 3 FDA-approved treatments that target the disease biology, but coverage by Medicare is limited, and 5 FDA-approved Alzheimer's drugs that treat symptoms. Non-pharmacological treatments target BPSD, which are behavioral and psychological symptoms of dementia like agitation, aggression, depression, and psychosis. These are treated through care in an assisted living/nursing facility 

* Disease progression & prognosis: Alzheimer’s can be a slow but gruelling progression that lasts up to several years. The rate of degeneration is unique for every patient and can be severe or slowly incremental. The prognosis is usually around 3 to 11 years but some individuals have lived up to 20 years after their diagnosis. Eventually, the disease will affect almost every part of the brain and result in the loss of ability to communicate, to be physically adept, and to function independently.  

* Continuum of care providers: There are a wide variety of providers that can help people who have been diagnosed with Alzheimer's. Primary care, neurologists, and psychiatrists can help with diagnosis and management of symptoms. Nurses, home health aides, and speech, physical, and occupational therapists can help patients with aspects of their daily lives. Some other providers include neuropsychologists, who are doctors who specialize in the study of cognition and behavior. Some geriatricians focus on treating elderly patients, which is relevant because more people with Alzheimer's are 65 years or older. 

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology): Alzheimer’s has primarily been linked to the buildup of amyloid-beta42 in plaques on the synapses of neurons and the tangling of abnormal phosphorylated tau proteins. The abnormal occurrences of those biomarkers lead to cell death and degeneration within the brain, causing dementia and severe cognitive decline. 

* Clinical Trials/next-gen therapies: There are around 466 clinical trials currently going on with the help of the National Institute on Aging (NIA). The institute is involved with clinical drug developments that are focused on amyloids, synaptic plasticity, inflammation, vasculature, multitarget, oxidative stress, receptors, and circadian rhythm. Non-drug clinical trials are involved with better understanding the connection between Alzheimer's and Exercise, neurostimulation, cognitive training, sleep, combination therapy, stress reduction/midfulness, diet/supplements, and social engagement. Other clinical trials focus on understanding disease processes, diagnostic tools, assessments, imaging studies, and treatments for neuropsychiatric symptoms.

## Data Analysis: 

We analyzed our data in a variety of ways; we first compared the published lifetime risk percentages based on APOE genotype with the percentage of diseased individuals per genotpye in our data. 

First, we created a patient class and assigned attributes based on the data from the csv dataset. We then created subsets of patients to sort by the specific attributes we wanted to study, like by APOE genotype.

Our first graph was a bar graph that displayed the prevalence of dementia for each APOE genotype compared to the reported lifetime-risk range for that genotype. For that graph, we used a chi square test to check whether there is truly an association between the APOE genotype and dementia status. 

```python
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import pandas as pd  

# loading the dataset 
df = pd.read_csv("Metadata and Protein Data for Module 1.csv")

#calculating the observed dementia prevalence for each APOE genotype
apoe_genotypes = ['3_3', '3_4', '4_4']
observed_prevalence = []
for genotype in apoe_genotypes:
    genotype_data = df[df['APOE Genotype'] == genotype]
    dementia_count = (genotype_data['Cognitive Status'] == 'Dementia').sum()
    total_count = len(genotype_data)
    prevalence = dementia_count/total_count * 100
    observed_prevalence.append(prevalence)

print(observed_prevalence)

#chi square test 
apoe_data = df[df['APOE Genotype'].isin(['3_3', '3_4', '4_4'])]
contingency_table = pd.crosstab(
    apoe_data['APOE Genotype'],
    apoe_data['Cognitive Status'] == 'Dementia'
)
print("Contingency Table: ")
print(contingency_table)
chi2, p_value, degrees_of_freedom, expected = stats.chi2_contingency(
    contingency_table
)
print("\nChi-square test results:")
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", p_value)

#creating bar graph that shows apoe genotype vs lifetime dementia risk %
#lifetime risk ranges from internet source 
apoe_genotypes = ['3_3', '3_4', '4_4']
risk_lower = np.array([10, 20, 30])
risk_upper = np.array([15, 25, 55])
risk_mean = (risk_lower + risk_upper)/2 #calculating midpoint for each risk range 
#calculating error bars for risk ranges
error_lower = risk_mean - risk_lower
error_upper = risk_upper - risk_mean
yerr = [error_lower, error_upper]

# Create bar graphs with groups
x = np.arange(len(apoe_genotypes))
width = 0.35

plt.bar(
    x - width/2,
    observed_prevalence,
    width,
    label="Observed dementia prevalence"
)

plt.bar(
    x + width/2,
    risk_mean,
    width,
    yerr=yerr,
    capsize=8,
    label="Reported lifetime risk"
)

plt.xticks(x, apoe_genotypes)
plt.xlabel("APOE Genotype")
plt.ylabel("Dementia Risk/Prevalence (%)")
plt.title("Observed Dementia Prevalence vs. Reported Lifetime Risk by APOE Genotype")
plt.legend()
plt.text(
    0.25, 0.78, 
    f"Chi-square p-value = {p_value:.4f}",
    transform =plt.gca().transAxes,
    ha='center'
)

plt.show()
#Used chat for code above to help plot 
```
