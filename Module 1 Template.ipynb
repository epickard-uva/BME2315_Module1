* Prevalence & incidence
The prevalence of a disease is the number of people currently living with the disease. The prevalence of Alzheimer's disease is over 55 million worldwide. The incidence rate is the number of new cases each year. The incidence of Alzheimer's disease is around 10 million worldwide. In the US, prevalence is around 7.4 million people, and incidence is around 1 million for individuals who are 65 years old and up.

* Risk factors (genetic, lifestyle) 
There are a variety of risk factors that influence somebody’s likelihood of developing Alzheimer's disease, including genetics, behaviors, habits, age, and family history. The largest factor is age, as the majority of people affected are 65 years or older. Having a direct family member with the disease also makes it more likely for it to develop. In terms of genetics, there are two types of genes that contribute to the development of the disease: deterministic and risk.  The Deterministic-type genes cause the disease but are responsible for less than 1% of cases. Some other factors include head injuries and heart problems such as diabetes or high blood pressure. Preventative lifestyle measures can also reduce risk, such as cognitive exercises, regular exercise, a healthy diet, not smoking, maintaining a healthy weight, and getting good sleep.  

Source: https://www.alz.org/alzheimers-dementia/what-is-alzheimers/causes-and-risk-factors

* Symptoms
The main symptom of Alzheimer's disease is memory loss that significantly impacts your daily life. Additional symptoms include difficulty planning, doing calculations, and concentrating on tasks. People’s perception of time and place can be altered as well. Mood changes may occur along with withdrawal from typical activities. Some other symptoms involve judgment becoming poor, trouble with speaking or conversating, and decreased ability to keep track of things.  A couple more symptoms involve vision being altered and judgment becoming poor. 

Source: https://www.alz.org/professionals/professional-providers/dementia_care_practice_recommendations

* Standard of care treatments (& reimbursement) 
Treatments of Alzheimer's involve different factors, including pharmacological treatment and non-pharmacological treatments. In terms of pharmacological options, there are 3 FDA-approved treatments that target the disease biology, but coverage by Medicare is limited, and 5 FDA-approved Alzheimer's drugs that treat symptoms. Non-pharmacological treatments target BPSD, which are behavioral and psychological symptoms of dementia like agitation, aggression, depression, and psychosis. These are treated through care in an assisted living/nursing facility

* Continuum of care providers
There are a wide variety of providers that can help people who have been diagnosed with Alzheimer's. Primary care, neurologists, and psychiatrists can help with diagnosis and management of symptoms. Nurses, home health aides, and speech, physical, and occupational therapists can help patients with aspects of their daily lives. Some other providers include neuropsychologists, who are doctors who specialize in the study of cognition and behavior. Some geriatricians focus on treating elderly patients, which is relevant because more people with Alzheimer's are 65 years or older. 

Source: https://www.alzheimers.gov/professionals/health-care-providers

* Clinical Trials/next-gen therapies
There are around 466 clinical trials currently going on with the help of the National Institute on Aging (NIA). The institute is involved with clinical drug developments that are focused on amyloids, synaptic plasticity, inflammation, vasculature, multitarget, oxidative stress, receptors, and circadian rhythm. Non-drug clinical trials are involved with better understanding the connection between Alzheimer's and Exercise, neurostimulation, cognitive training, sleep, combination therapy, stress reduction/midfulness, diet/supplements, and social engagement. Other clinical trials focus on understanding disease processes, diagnostic tools, assessments, imaging studies, and treatments for neuropsychiatric symptoms.
